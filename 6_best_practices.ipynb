## 6.1 Testing Python Code with pytest
- 将模型逻辑与云客户端分离，并通过快速单元测试覆盖服务。
- 最佳实践部分从第4模块中的流式模型服务开始。第一步是在不部署整个云架构的情况下，使其行为可测试。
