# 一、实验跟踪
## 实验 Experiment
- 实验是更大规模的建模活动，例如尝试多种算法和超参数进行出租车行程时长预测。
- 每次试验都算一次运行。
- 一次运行可以记录参数、指标和元数据，也可以保存文件和训练好的模型。
## 跟踪 Tracking
- 记录在构建模型过程中发生变化的信息。
- 帮助我们重现结果、比较试验，并回归早期决策，而非依赖电子表格或记忆。
## 参数 Parameters
- 包括模型的配置、训练数据路径和验证数据路径等值。
## 指标 Metrics
- 包括在训练集或验证集上计算的RMSE等度量。
## 构建产物   Artifacts
- 是指我们希望随运行一起保留的文件，例如图表、序列化模型或报告。

# 二、追踪运行的原因   
## 1、 可重复性：可以查看是哪段代码、数据和参数产生了某个结果。  
## 2、 组织性：团队成员可以找到实验，并理解哪些内容发生了变化。  
## 3、 优化性：可以比较多次试验，选择有依据的模型。

# 三、Start with MLflow
- MLflow 是一个用于机器学习生命周期的开源 Python 包。
- 其追踪模块可将实验组织为运行，而模型注册表则帮助我们在选定候选模型后进行管理。
- 你可以先从本地的 MLflow 配置开始。当多人需要共享实验运行 runs和构建产物artifacts时，远程跟踪服务器会变得有用，但首次实验时并不需要它。
