# MLOps 成熟度模型

## 0级：禁止MLOps自动化
- 在0级时，我们在没有完善流程、实验跟踪或模型元数据的jupyter notebook中工作。
- 数据科学家可能会把jupyter notebook交给工程师，工程师再为某项服务重新实现。这个层级通常足以让概念进行早期验证。

## 第一层：无MLOps的DevOps
- 在第一层，团队采用常规的软件工程实践。发布自动化，测试覆盖服务，CI/CD 执行检查，操作指标显示服务是否可用。
- 团队仍然缺乏机器学习专用的跟踪功能、易于复制的模型，以及数据科学家与工程师之间清晰的联系。

## 第二阶段：自动化training
- 在2级时，类似命令的指令可以运行训练过程而无需打开jupyter notebook。
- 脚本接受训练月份等参数，记录实验，并识别当前生产中的模型。
- 部署仍可能采用手动，但培训过程可重复且摩擦较低。

## 三级：自动部署
- 在3级时，训练好的模型作为工作流程的一部分迁移到服务平台。
- 该平台可以公开模型端点，并可能支持比较两个模型版本的A/B测试。
- 监控成为部署过程的一部分，因为我们需要知道模型在开始服务流量后是否正常工作。

## 第四级：全面MLOps自动化
- 在4级时，监控可以触发剩余的过程。
- 系统检测模型漂移或性能下降。
- 它开始gtraining，评估新模型，部署，并在结果更好时向用户推广。
