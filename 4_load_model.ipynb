```python
# 告诉 MLflow，实验数据存在 http://127.0.0.1:5000 这个服务器上。后面加载模型时，会从这里找。
import mlflow
TRACKING_URL = "http://127.0.0.1:5000"
mlflow.set_tracking_uri(TRACKING_URL)
```

```python
run_id = 'adef29c02b1d4dd781978ba4e90b5ca4'
logged_model = f'runs:/{run_id}/models_mlflow'

loaded_model = mlflow.pyfunc.load_model(logged_model)
```

```python
request = {
    'tpep_pickup_datetime': '2021-01-01 00:15:56',
    'PULocationID': 43,
    'DOLocationID': 151,
    'passenger_count': 1.0,
    'trip_distance': 1.01
}
```

```python
features = {}
features['PU_DO'] = '%s_%s' % (request['PULocationID'], request['DOLocationID'])
features['trip_distance'] = request['trip_distance']
```

```python
local_path = mlflow.artifacts.download_artifacts(
    run_id=run_id,
    artifact_path="preprocessor/preprocessor.b"
)
```

```python
import pickle
with open(local_path, "rb") as f:
    dv = pickle.load(f)

X = dv.transform([features])
```

```python
loaded_model.predict(X)
```
